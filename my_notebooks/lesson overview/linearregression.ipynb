## 📊 Explication Détaillée des Coefficients (EN FRANÇAIS)

### Qu'est-ce qu'un coefficient ?
Un **coefficient** dans une régression linéaire représente l'**impact** d'une variable indépendante sur la variable dépendante. C'est le **taux de changement**.

### Notre modèle :
**Prix de fermeture (Close) = Intercept + β₁×Open + β₂×High + β₃×Low + β₄×Volume**

---

### 🔍 Interprétation de chaque coefficient :

#### **1. Coefficient de "Open" (prix d'ouverture)**
- **Signification** : Pour chaque augmentation de **1 unité** du prix d'ouverture, le prix de fermeture augmentera d'environ **β₁ unités** (en gardant les autres variables constantes).
- **Exemple concret** :
  - Si β₁ = 0.95, cela signifie que si le stock ouvre à 100€ au lieu de 99€, on s'attend à ce que le prix de fermeture soit environ **0.95€ plus haut**.
  - Les stocks qui ouvrent plus cher ferment généralement aussi plus cher (corrélation positive).

#### **2. Coefficient de "High" (prix le plus haut)**
- **Signification** : Pour chaque augmentation de **1 unité** du prix maximum du jour, le prix de fermeture augmentera d'environ **β₂ unités**.
- **Exemple concret** :
  - Si β₂ = 0.85, cela signifie que si le prix maximum du jour monte de 1€ supplémentaire, le prix de fermeture augmentera en moyenne de **0.85€**.
  - C'est logique : un jour où le stock monte haut tend à fermer plus haut.

#### **3. Coefficient de "Low" (prix le plus bas)**
- **Signification** : Pour chaque augmentation de **1 unité** du prix minimum du jour, le prix de fermeture augmentera d'environ **β₃ unités**.
- **Exemple concret** :
  - Si β₃ = -0.40, cela signifie une **relation inverse** (coefficient négatif) : si le prix minimum baisse davantage, cela indique une journée volatile, et le prix de fermeture sera probablement plus bas.
  - Un coefficient négatif ne signifie pas que le phénomène est mauvais, seulement que la relation est **inversée**.

#### **4. Coefficient de "Volume" (volume de transactions)**
- **Signification** : Pour chaque augmentation de **1 unité** du volume (nombre d'actions échangées), le prix de fermeture changera d'environ **β₄ unités**.
- **Exemple concret** :
  - Si β₄ = 0.000001 (très petit), cela signifie que le volume a une **influence mineure** sur le prix de fermeture.
  - Volume élevé = plus d'achats/ventes, mais cela n'influence pas fortement où le stock ferme.

---

### 📐 L'Intercept (Ordonnée à l'origine)
- **Signification** : C'est la valeur prédite du prix de fermeture quand **toutes les variables indépendantes sont à 0**.
- Dans la pratique : c'est le "prix de base" du modèle.
- **Exemple** : Si intercept = 2.5€, cela veut dire qu'à partir d'une "base" de 2.5€, on ajoute/retranche les effets de chaque variable.

---

### 💡 Comment utiliser les coefficients pour faire des prédictions ?

**Exemple avec des chiffres concrets** :
Si nos coefficients sont :
- Intercept = 2.00
- β₁ (Open) = 0.90
- β₂ (High) = 0.80
- β₃ (Low) = -0.30
- β₄ (Volume) = 0.00001

Et qu'un jour nous avons :
- Open = 100€
- High = 105€
- Low = 98€
- Volume = 1,000,000 actions

**Prédiction du Close** :
```
Close = 2.00 + (0.90 × 100) + (0.80 × 105) + (-0.30 × 98) + (0.00001 × 1,000,000)
      = 2.00 + 90 + 84 + (-29.4) + 10
      = 156.60€
```

---

### ✅ Ce que nous pouvons conclure :

1. **Open, High impactent positivement** → Plus le stock ouvre/monte haut, plus il ferme haut.
2. **Low a un impact négatif** → Plus le prix baisse dans la journée, plus il ferme bas (logique !).
3. **Volume a peu d'impact** → Le nombre de transactions n'influence pas beaucoup le prix de fermeture.
4. **La magnitude du coefficient** → Plus grande |β|, plus l'impact est fort.

## 🎯 Explication Détaillée de l'Évaluation du Modèle (EN FRANÇAIS)

### Pourquoi évaluer un modèle ?
Après avoir entraîné un modèle de régression, nous devons vérifier **si le modèle fonctionne bien** sur des données qu'il n'a jamais vues (données de test). C'est comme passer un examen : si tu as étudié trop longtemps un seul sujet, tu pourrais bien le maîtriser mais échouer sur d'autres. C'est le problème du **surapprentissage (overfitting)**.

---

### 📊 Métrique 1 : R-squared (R²) - Le Coefficient de Détermination

#### **Définition Simple**
**R² mesure la proportion de la variance dans les données que le modèle explique.**

C'est comme demander : *"Quelle fraction de la variation du prix de fermeture mon modèle peut-il prédire correctement ?"*

#### **Formule (conceptuelle)**
```
R² = 1 - (Erreur du modèle / Variance totale)
```

#### **Interprétation**
- **R² = 0.95** → Le modèle explique 95% de la variation → **EXCELLENT** ✅
- **R² = 0.80** → Le modèle explique 80% de la variation → **BON** ✅
- **R² = 0.50** → Le modèle explique 50% de la variation → **MOYEN** ⚠️
- **R² = 0.20** → Le modèle explique seulement 20% → **MAUVAIS** ❌
- **R² < 0** → Le modèle est pire qu'une simple moyenne → **TRÈS MAUVAIS** 🔴

#### **Exemple Concret**
Imaginez que les prix des stocks oscillent entre 95€ et 105€ chaque jour.
- Si votre modèle prédit correctement 95% de ces oscillations (R² = 0.95), vous êtes très satisfait.
- Si votre modèle ne capture que 50% des oscillations (R² = 0.50), il vous manque beaucoup d'informations.

---

### 📈 Métrique 2 : MSE (Mean Squared Error) - Erreur Quadratique Moyenne

#### **Définition Simple**
**MSE mesure la magnitude moyenne des erreurs de prédiction, au carré.**

C'est la distance moyenne entre les valeurs réelles et les valeurs prédites.

#### **Formule (conceptuelle)**
```
MSE = (1/n) × Σ(Valeur réelle - Valeur prédite)²
```

#### **Pourquoi élever au carré ?**
- Cela pénalise **fortement les grandes erreurs** (erreur de 10€ → 100 au carré)
- Les erreurs petites sont peu pénalisées (erreur de 1€ → 1 au carré)

#### **Interprétation**
- **MSE petit** (ex: 0.5) → Le modèle fait des petites erreurs → **BON** ✅
- **MSE moyen** (ex: 10) → Le modèle fait des erreurs modérées → **ACCEPTABLE** ⚠️
- **MSE grand** (ex: 100) → Le modèle fait de grandes erreurs → **MAUVAIS** ❌

**IMPORTANT** : MSE dépend des unités de votre variable (euros, milliers d'euros, etc.). Ce n'est pas facile à comparer entre deux datasets différents.

#### **Exemple Concret**
Vous prédisez les prix de fermeture :
- **Valeur réelle** : 100€, **Prédiction** : 98€ → Erreur = 2€ → (2)² = 4
- **Valeur réelle** : 105€, **Prédiction** : 107€ → Erreur = -2€ → (-2)² = 4
- **Valeur réelle** : 110€, **Prédiction** : 105€ → Erreur = 5€ → (5)² = 25

**MSE = (4 + 4 + 25) / 3 = 11**

Ce MSE de 11 signifie qu'en moyenne, votre modèle se trompe de √11 ≈ 3.3€.

---

### 🔗 Relation entre R² et MSE

| Scénario | R² | MSE | Signification |
|----------|----|----|---|
| Modèle parfait | 1.0 | 0 | Prédictions exactes |
| Très bon modèle | 0.85-0.95 | Très bas | Peu d'erreurs |
| Bon modèle | 0.70-0.85 | Bas | Quelques erreurs |
| Modèle acceptable | 0.50-0.70 | Moyen | Erreurs modérées |
| Mauvais modèle | <0.50 | Haut | Beaucoup d'erreurs |

---

### 💡 Alternative Utile : RMSE (Root Mean Squared Error)

**Formule** :
```
RMSE = √MSE
```

**Avantage** : RMSE est dans la **même unité que votre variable** (euros dans notre cas).

**Exemple** :
- Si MSE = 11, alors RMSE = √11 ≈ 3.3€
- Cela signifie : "En moyenne, mon modèle se trompe de 3.3€"
- C'est beaucoup plus facile à comprendre que MSE = 11 !

---

### ✅ Checklist : Comment Interpréter l'Évaluation ?

1. **Vérifier R²** :
   - Si R² > 0.7 → ✅ Le modèle capture bien la tendance
   - Si R² < 0.5 → ❌ Le modèle ne capture pas assez de variance

2. **Vérifier MSE/RMSE** :
   - Comparer avec l'écart-type des données réelles
   - Si RMSE << écart-type → ✅ Le modèle est bien
   - Si RMSE ≈ écart-type → ❌ Le modèle n'est pas mieux qu'une moyenne

3. **Visualiser les résidus** (erreurs) :
   - Les erreurs doivent être **aléatoires**, sans pattern
   - Si les erreurs montrent une tendance → Le modèle omet une variable importante

4. **Vérifier l'overfitting** :
   - Comparer R² sur **données d'entraînement** vs **données de test**
   - Si R²_train >> R²_test → ⚠️ Overfitting (le modèle a mémorisé les données)
   - Si R²_train ≈ R²_test → ✅ Bon équilibre
