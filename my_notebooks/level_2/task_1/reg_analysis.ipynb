<p>Tools: Python, scikit-learn, pandas</p>









```python
import sys
from pathlib import Path

import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn import metrics
from sklearn.model_selection import train_test_split

PROJECT_ROOT = Path.cwd().parent.parent.parent
sys.path.insert(0, str(PROJECT_ROOT))



CLEAN_DIR = PROJECT_ROOT / "data/cleaned_data"
CLEAN_DIR.mkdir(parents=True, exist_ok=True)

print(f"Projet : {PROJECT_ROOT}")
print(f"Cleaned files : {len(list(CLEAN_DIR.rglob('*.csv')))}")
```

<h4>The dataset: Stock prices</h4>
<p>
    <h4><strong><li>Data regression analysis to predict the "close" value based on "open", "high", "low", "volume"</li></strong></h4>
    We split the data to train the model and validate the training
</p>

```python
#Load the data
stock = pd.read_csv(CLEAN_DIR / "c_stock.csv")
display(stock.head(5))
```

```python
features = ['open','high','low','volume']
X = stock[features]
y = stock.close

#Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=14)

#Shape of the X_train and X_test
print(f'X_train shape:{X_train.shape}\nX_test shape:{X_test.shape}')

```

```python
#Create the linear model
linear_model = LinearRegression()
linear_model.fit(X_train,y_train)
```

<p>
    <h4><strong><li>Coefficient: calculate and interprete</li></strong></h4>
</p>

```python
coef = linear_model.coef_
```

```python
# Coefficient interpretation for the linear regression model

intercept = linear_model.intercept_
print(f"Estimated close = {intercept:.4f} + {coef[0]:.4f}*open + {coef[1]:.4f}*high + {coef[2]:.4f}*low + {coef[3]:.4f}*volume")

print("\nInterpretation of coefficients (holding the other variables constant):")
for feature, beta in zip(X.columns, coef):
    if abs(beta) < 1e-8:
        print(f"- {feature}: coefficient ≈ 0, so this feature has almost no linear effect on close in this model.")
    else:
        print(f"- {feature}: a 1-unit increase in {feature} is associated with an approximate {beta:.6f}-unit change in close, holding the other variables constant.")

# Optional: show coefficients in a neat table
coef_df = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": coef
})
display(coef_df.round(6))
```

<p>
    <h4><strong><li>Predictions from our model</li></strong></h4>
</p>

```python
y_predicted = linear_model.predict(X_test)
display(y_predicted)
```

```python
display(y_test.head(5))
display(y_predicted[0:5])
```

<p>
    <h4><strong><li>Evaluate the model using R-squared and mean squared error (MSE)</li></strong></h4>
</p>

```python
# Make predictions on test data
y_predicted = linear_model.predict(X_test)

# Calculate metrics
r2 = linear_model.score(X_test, y_test)
mse = metrics.mean_squared_error(y_test, y_predicted)
rmse = np.sqrt(mse)

# Display results
print("MODEL EVALUATION")
print("=" * 40)
print(f"R-squared (R²):  {r2:.4f}")
print(f"Mean Squared Error (MSE):  {mse:.6f}")
print(f"Root Mean Squared Error (RMSE):  {rmse:.6f}")
print("=" * 40)
```

```python
# Interpret the results
print("\nINTERPRETATION")
print("=" * 40)

# Interpret R-squared
if r2 >= 0.9:
    print(f"✅ R² = {r2:.4f}: EXCELLENT")
    print("   The model explains >90% of variance")
elif r2 >= 0.8:
    print(f"✅ R² = {r2:.4f}: GOOD")
    print("   The model explains 80-90% of variance")
elif r2 >= 0.7:
    print(f"⚠️  R² = {r2:.4f}: ACCEPTABLE")
    print("   The model explains 70-80% of variance")
elif r2 >= 0.5:
    print(f"⚠️  R² = {r2:.4f}: WEAK")
    print("   The model explains 50-70% of variance")
else:
    print(f"❌ R² = {r2:.4f}: POOR")
    print("   The model explains <50% of variance")

# Interpret MSE
print(f"\nMSE = {mse:.6f}")
print("   Lower values = Better predictions")
print("=" * 40)
```
