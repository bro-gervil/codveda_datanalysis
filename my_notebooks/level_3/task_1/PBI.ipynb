<h1><strong>Task 2:  Building Dashboards with Power BI/Tableau</strong></h1>
<p><strong>Description:</strong> Create interactive dashboards using a data visualization tool like Power BI or Tableau to present insights from a dataset.</p>

<h3><strong> Objectives:</strong></h3> 
<ul>
    <li>Import and clean the dataset in Power BI/Tableau.</li>
    <li>Create interactive visualizations (e.g., bar charts, line graphs, maps).</li>
    <li>Set up filters and slicers for interactive exploration.</li>
    <li>Publish the dashboard and share it with others.</li>
</ul>
<p>Tools: Power BI or Tableau.</p>
